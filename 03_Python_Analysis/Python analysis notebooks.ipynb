```python

import pandas as pd
from pathlib import Path

# Use the validated delivery data
deliveries = pd.read_csv(
    r"C:\Users\reddy\OneDrive\Documents\T20_World_Cup_Cricket_Analytics\02_Data\Cleaned\deliveries.csv"
)

# Exclude Super Overs from regular innings statistics
regular = deliveries[
    ~deliveries["innings_name"].str.contains(
        "Super Over", case=False, na=False
    )
].copy()

# Ensure numeric data types
for col in ["batter_runs", "wide_runs", "no_ball_runs"]:
    regular[col] = pd.to_numeric(
        regular[col], errors="coerce"
    ).fillna(0)

# Balls faced: wides do not count as balls faced
regular["balls_faced"] = (
    regular["wide_runs"] == 0
).astype(int)

# Boundaries credited to the batter
regular["fours"] = (regular["batter_runs"] == 4).astype(int)
regular["sixes"] = (regular["batter_runs"] == 6).astype(int)

# Aggregate player batting performance
batting_summary = (
    regular.groupby("batter")
    .agg(
        runs=("batter_runs", "sum"),
        balls_faced=("balls_faced", "sum"),
        fours=("fours", "sum"),
        sixes=("sixes", "sum")
    )
    .reset_index()
)

batting_summary["strike_rate"] = (
    batting_summary["runs"] /
    batting_summary["balls_faced"].replace(0, pd.NA) * 100
).round(2)

batting_summary = batting_summary.sort_values(
    ["runs", "strike_rate"], ascending=[False, False]
)

# Save processed output
project_folder = Path(
    r"C:\Users\reddy\OneDrive\Documents\T20_World_Cup_Cricket_Analytics"
)

processed_folder = project_folder / "02_Data" / "Processed"
processed_folder.mkdir(parents=True, exist_ok=True)

batting_summary.to_csv(
    processed_folder / "batting_summary.csv",
    index=False
)

print("Regular-innings deliveries:", len(regular))
print("Players in batting summary:", len(batting_summary))
print("\nTop 10 batters:")
display(batting_summary.head(10))
```

```python

from pathlib import Path
from collections import Counter
import yaml

dismissal_types = Counter()
files_processed = 0
files_failed = []
total_dismissals = 0
total_bowler_credited = 0

for path in extracted_folder.glob("*.yaml"):
    try:
        with path.open("r", encoding="utf-8") as file:
            match = yaml.safe_load(file)

        if not isinstance(match, dict):
            continue

        files_processed += 1

        for innings_entry in match.get("innings", []):
            if not isinstance(innings_entry, dict):
                continue

            for innings_name, innings_data in innings_entry.items():
                if not isinstance(innings_data, dict):
                    continue

                for delivery_over in innings_data.get("deliveries", []):
                    if not isinstance(delivery_over, dict):
                        continue

                    for ball_label, delivery in delivery_over.items():
                        if not isinstance(delivery, dict):
                            continue

                        wicket = delivery.get("wicket")

                        if not isinstance(wicket, dict):
                            continue

                        kind = wicket.get("kind", "unknown")
                        dismissal_types[kind] += 1
                        total_dismissals += 1

                        # These dismissal types are not credited
                        # to the bowler in standard cricket scoring.
                        non_bowler_dismissals = {
                            "retired hurt",
                            "obstructing the field",
                            "run out",
                            "retired out",
                            "handled the ball",
                        }

                        if kind not in non_bowler_dismissals:
                            total_bowler_credited += 1

    except Exception as error:
        files_failed.append((path.name, str(error)))

print("Files processed:", files_processed)
print("Files failed:", len(files_failed))
print("Total dismissal records:", total_dismissals)
print("Bowler-credited dismissals (provisional):",
      total_bowler_credited)

print("\nDismissal types:")
for kind, count in dismissal_types.most_common():
    print(f"{kind}: {count}")

if files_failed:
    print("\nFailed files (first 10):")
    for item in files_failed[:10]:
        print(item)
```

```python

import pandas as pd
import yaml
from collections import defaultdict

bowling_stats = defaultdict(lambda: {
    "wickets": 0,
    "runs_conceded": 0,
    "legal_balls": 0,
    "matches": set()
})

# Dismissals that are not credited to the bowler
non_bowler_dismissals = {
    "run out",
    "retired hurt",
    "retired out",
    "obstructing the field",
}

for path in extracted_folder.glob("*.yaml"):
    with path.open("r", encoding="utf-8") as file:
        match = yaml.safe_load(file)

    if not isinstance(match, dict):
        continue

    match_id = path.stem

    for innings_entry in match.get("innings", []):
        if not isinstance(innings_entry, dict):
            continue

        for innings_name, innings_data in innings_entry.items():
            if not isinstance(innings_data, dict):
                continue

            for delivery_over in innings_data.get("deliveries", []):
                if not isinstance(delivery_over, dict):
                    continue

                for ball_label, delivery in delivery_over.items():
                    if not isinstance(delivery, dict):
                        continue

                    bowler = delivery.get("bowler")
                    if not bowler:
                        continue

                    stats = bowling_stats[bowler]
                    stats["matches"].add(match_id)

                    runs = delivery.get("runs", {})
                    extras = delivery.get("extras", {})

                    # Runs charged to the bowler:
                    # exclude byes and leg-byes.
                    bowler_runs = runs.get("total", 0)
                    bowler_runs -= extras.get("byes", 0)
                    bowler_runs -= extras.get("legbyes", 0)
                    bowler_runs -= extras.get("penalty", 0)

                    stats["runs_conceded"] += bowler_runs

                    # Legal deliveries exclude wides and no-balls.
                    if "wides" not in extras and "noballs" not in extras:
                        stats["legal_balls"] += 1

                    wicket = delivery.get("wicket", {})
                    if isinstance(wicket, dict) and wicket:
                        kind = wicket.get("kind", "").lower()

                        if kind and kind not in non_bowler_dismissals:
                            stats["wickets"] += 1

bowling_summary = pd.DataFrame([
    {
        "bowler": bowler,
        "matches": len(stats["matches"]),
        "wickets": stats["wickets"],
        "runs_conceded": stats["runs_conceded"],
        "legal_balls": stats["legal_balls"],
    }
    for bowler, stats in bowling_stats.items()
])

bowling_summary["overs"] = (
    bowling_summary["legal_balls"] // 6
    + (bowling_summary["legal_balls"] % 6) / 10
)

bowling_summary["economy"] = (
    bowling_summary["runs_conceded"]
    / (bowling_summary["legal_balls"] / 6).replace(0, float("nan"))
)

bowling_summary["bowling_average"] = (
    bowling_summary["runs_conceded"]
    / bowling_summary["wickets"].replace(0, float("nan"))
)

bowling_summary = bowling_summary.sort_values(
    ["wickets", "economy"],
    ascending=[False, True]
)

display(bowling_summary.head(15))
print("Unique bowlers:", bowling_summary["bowler"].nunique())
```
